**📌 Project Overview**

This project focuses on performing Exploratory Data Analysis (EDA) on a dataset containing information about used cars listed on the CarDekho platform. The objective is to understand car characteristics, pricing patterns, ownership details, fuel types, transmission types, mileage, engine specifications, and other factors associated with car prices.

The analysis explores car listings across different brands, fuel types, transmission types, ownership categories, and price ranges to identify meaningful patterns, relationships, trends, and unusual values within the dataset.

---

**🎯 Objectives**

  *   Understand the structure and characteristics of the dataset.
  *   Perform data quality checks and data cleaning.
  *   Analyze missing and duplicate values.
  *   Study the distribution of numerical and categorical variables.
  *   Detect and analyze outliers using IQR and Z-score methods.
  *   Analyze car prices across different brands and categories.
  *   Explore the relationship between car price and numerical features.
  *   Analyze fuel type and transmission patterns.
  *   Study ownership patterns across different car categories.
  *   Examine the relationship between engine, mileage, power, and price.
  *   Perform correlation analysis among numerical variables.
  *   Generate meaningful insights using data visualization.

---

**📊 Dataset**

The dataset contains approximately 37,813 used-car listings from the CarDekho platform.

It includes information related to:

Car Name, Brand, Model, Price, Year, Fuel Type, Transmission, Owner, Mileage, Engine, Power, Seats, and other car specifications.

```python
# Separate numerical and categorical columns
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Fill numerical missing values with median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

**10. Outlier Analysis**


```python
#IQR Method — Calculate Outliers
# Variables for outlier analysis
outlier_columns = [
    'Price_Lakhs',
    'km',
    'Car Age',
    'Max Power Delivered',
    'Max Torque Delivered',
    'Valves per Cylinder'
]

# Calculate IQR outliers
for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f'\n{col}')
    print(f'Q1: {Q1:.2f}')
    print(f'Q3: {Q3:.2f}')
    print(f'IQR: {IQR:.2f}')
    print(f'Lower Bound: {lower_bound:.2f}')
    print(f'Upper Bound: {upper_bound:.2f}')
    print(f'Number of Outliers: {len(outliers)}')
```

```python
#Z-Score Method
for col in outlier_columns:

    z_scores = np.abs(stats.zscore(df[col]))

    outliers = df[z_scores > 3]

    print(f'{col}: {len(outliers)} Z-score outliers')
```

```python
#Compare IQR and Z-score
outlier_summary = []

for col in outlier_columns:

    # IQR
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    iqr_count = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    # Z-score
    z_scores = np.abs(stats.zscore(df[col]))
    z_count = (z_scores > 3).sum()

    outlier_summary.append([
        col,
        iqr_count,
        z_count
    ])

outlier_summary = pd.DataFrame(
    outlier_summary,
    columns=[
        'Variable',
        'IQR Outliers',
        'Z-Score Outliers'
    ]
)

outlier_summary
```

```python
#Create the Correlation Dataset
# Select important numerical variables for correlation analysis

correlation_columns = [
    'Price_Lakhs',
    'Car Age',
    'km',
    'Max Power Delivered',
    'Max Power At',
    'Max Torque Delivered',
    'Max Torque At'
]

correlation_df = df[correlation_columns]

correlation_df.head()
```

```python
#Price Correlation
# Correlation of variables with Price

price_correlation = correlation_matrix['Price_Lakhs'].sort_values(
    ascending=False
)

price_correlation
```

```python
#Find Strong Correlations
# Identify strong correlations

strong_correlations = []

for col1 in correlation_matrix.columns:
    for col2 in correlation_matrix.columns:

        if col1 != col2:
            correlation_value = correlation_matrix.loc[col1, col2]

            if abs(correlation_value) >= 0.5:
                strong_correlations.append(
                    [col1, col2, correlation_value]
                )

strong_correlations_df = pd.DataFrame(
    strong_correlations,
    columns=['Variable 1', 'Variable 2', 'Correlation']
)

strong_correlations_df
```
